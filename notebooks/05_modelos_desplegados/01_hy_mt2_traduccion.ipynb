# Hy-MT2: traducción

Ejemplo mínimo de cómo traducir con Hy-MT2-1.8B por OpenRouter con el SDK `openrouter`.

**Antes de correr:** `OPENROUTER_API_KEY` en el `.env` de la raíz del repo. Se paga por token (centavos por miles de reseñas).

OpenRouter lo sirve Tencent en fp8 y solo acepta `temperature`: no se pueden fijar `top_p`, `top_k` ni `repetition_penalty`.

```python
import os
from pathlib import Path

from dotenv import load_dotenv
from openrouter import OpenRouter

ROOT = Path("../..")  # raíz del repo, relativa a este notebook
load_dotenv(ROOT / ".env")

client = OpenRouter(api_key=os.environ["OPENROUTER_API_KEY"])
```

```python
def traducir(texto: str, idioma: str = "English") -> str:
    # prompt recomendado por el modelo
    prompt = (
        f"Translate the following text into {idioma}. Note that you should "
        "only output the translated result without any additional explanation:\n\n" + texto
    )
    res = client.chat.send(
        model="tencent/hy-mt2-1.8b",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=2048,
        temperature=0.7,
    )
    return res.choices[0].message.content.strip()
```

```python
traducir("La comida estaba riquísima, pero tardaron una hora en traerla.")
```
