# Qwen: chat

Ejemplo mínimo de cómo llamar a Qwen3.5-9B por OpenRouter con el SDK `openrouter`.

**Antes de correr:** `OPENROUTER_API_KEY` en el `.env` de la raíz del repo. Se paga por token (centavos por miles de reseñas).

```python
import os
from pathlib import Path

from dotenv import load_dotenv
from openrouter import OpenRouter

ROOT = Path("../..")  # raíz del repo, relativa a este notebook
load_dotenv(ROOT / ".env")

client = OpenRouter(api_key=os.environ["OPENROUTER_API_KEY"])
```

```python
def chat(mensaje: str) -> str:
    res = client.chat.send(
        model="qwen/qwen3.5-9b",
        messages=[{"role": "user", "content": mensaje}],
        max_tokens=1024,
        temperature=0.7,
        reasoning={"enabled": False},
        provider={"quantizations": ["bf16"]},
    )
    return res.choices[0].message.content
```

```python
chat("Hola como estas?")
```
